```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
```

```python
df = pd.read_csv("Groceries_dataset.csv.zip")
df.head()
```

```python
basket = df.groupby(['Member_number'])['itemDescription'].apply(list).values.tolist()

te = TransactionEncoder()
te_array = te.fit(basket).transform(basket)
df_encoded = pd.DataFrame(te_array, columns=te.columns_)
df_encoded.head()
```

```python
frequent_items = apriori(df_encoded, min_support=0.03, use_colnames=True)
frequent_items.sort_values('support', ascending=False).head(10)
```

```python
rules = association_rules(frequent_items, metric="lift", min_threshold=1.0)
rules = rules.sort_values(['lift', 'confidence'], ascending=False)
rules.head()
```
